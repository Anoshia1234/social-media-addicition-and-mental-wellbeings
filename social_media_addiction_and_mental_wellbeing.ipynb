```python
# Count missing values per column
missing_counts = df.isnull().sum()
missing_filtered = missing_counts[missing_counts > 0]
print("Missing Values Per Column:\n", missing_filtered)
```

```python
# Visualize missing values before treatment
plt.figure(figsize=(10, 5))
sns.barplot(x=missing_filtered.index, y=missing_filtered.values, palette='viridis', hue=missing_filtered.index, legend=False)
plt.title('Missing Values Count Before Treatment')
plt.xlabel('Columns')
plt.ylabel('Missing Count')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# Missing values treatment before implementation
print("Total missing values before treatment:", df.isnull().sum().sum())
```

```python
# Fill numerical nulls with median
for col in df.select_dtypes(include=[np.number]).columns:
    df[col] = df[col].fillna(df[col].median())

# Fill categorical nulls with mode
for col in df.select_dtypes(include=['object']).columns:
    df[col] = df[col].fillna(df[col].mode()[0])

# Verification print
print("Total missing values after treatment:", df.isnull().sum().sum())
```

```python
# Check duplicate rows
duplicate_rows = df.duplicated().sum()
print(f"Total Duplicate Rows: {duplicate_rows}")
```

```python
# Check duplicate User IDs
duplicate_users = df['User_ID'].duplicated().sum()
print(f"Duplicate User IDs: {duplicate_users}")

if duplicate_rows == 0:
    print("Proof: No duplicate rows found in the dataset.")
else:
    df = df.drop_duplicates()
    print("Duplicate rows successfully removed.")
```

```python
# Box plots and histograms for key numeric columns
fig, axes = plt.subplots(1, 2, figsize=(10, 3))
sns.boxplot(x=df['Daily_Usage_Hours'], ax=axes[0], color='skyblue')
axes[0].set_title('Box Plot of Daily Usage Hours')

sns.histplot(df['Daily_Usage_Hours'], kde=True, ax=axes[1], color='skyblue')
axes[1].set_title('Histogram of Daily Usage Hours')
plt.tight_layout()
plt.show()
```

```python
# IQR Rule on Daily_Usage_Hours and Anxiety_Score
def clean_outliers(data, col):
    Q1 = data[col].quantile(0.25)
    Q3 = data[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return data[(data[col] >= lower) & (data[col] <= upper)]

df_cleaned = clean_outliers(df, 'Daily_Usage_Hours')
print(f"Rows before outlier filtering: {len(df)}, After filtering: {len(df_cleaned)}")
```

Explanation: Outliers were evaluated using the Interquartile Range (IQR) method. Extreme values in usage hours and psychological metrics represent genuine behavioral extremes (heavy users) rather than data entry errors, hence retained or safely managed for robust model/statistical integrity.
